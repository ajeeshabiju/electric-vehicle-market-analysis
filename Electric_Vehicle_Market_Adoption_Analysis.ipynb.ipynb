## Missing Value Treatment

Missing values were examined separately based on the nature and purpose of each variable.

Geographic identifier fields such as `postal_code`, `legislative_district`, and `2020_geoid` were retained because they provide useful geographic information. Missing observations in these fields were categorized as `"Unknown"` rather than removing the corresponding records.

The `electric_range` column was treated differently because it is a numerical analytical variable. Missing range values were not replaced with an arbitrary mean or median, as doing so could introduce artificial values and affect the statistical analysis. Instead, the missing observations were excluded only from analyses that specifically require electric-range values.

This approach preserves as many records as possible while avoiding inappropriate numerical imputation.

### 7.Outlier Detection

```python
Q1 = df["electric_range"].quantile(0.25)
Q3 = df["electric_range"].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df["electric_range"] < lower_bound) |
    (df["electric_range"] > upper_bound)
]

print("Number of outliers:", len(outliers))
```

## Feature Engineering

Feature 1 — Vehicle Age

```python
reference_year = df["model_year"].max()

df["vehicle_age"] = (
    reference_year - df["model_year"]
)
```

Feature 2 — Range Category

```python
df["range_category"] = pd.cut(
    df["electric_range"],
    bins=[-1, 100, 200, 300, 400, np.inf],
    labels=[
        "0-100",
        "101-200",
        "201-300",
        "301-400",
        "400+"
    ]
)
```

Feature 3 — Model Year Group

```python
df["model_year_group"] = pd.cut(
    df["model_year"],
    bins=[0, 2015, 2018, 2020, 2022, 2024, np.inf],
    labels=[
        "Before 2016",
        "2016-2018",
        "2019-2020",
        "2021-2022",
        "2023-2024",
        "2025+"
    ]
)
```

## Final Conclusion

This project analyzed 299,705 electric vehicle registration records using Python and performed a complete data analytics workflow involving data loading, data cleaning, preprocessing, feature engineering, exploratory data analysis, visualization, statistical analysis and insight generation.

The analysis shows that Battery Electric Vehicles form the majority of the observed EV population, while Tesla is the most represented manufacturer. Recent model years contain substantially more records than older model years, indicating strong representation of newer EV models within the dataset.

The geographic analysis reveals a major concentration of records in Washington State. Therefore, the dataset should not be interpreted as a nationally representative measure of EV adoption. Instead, it is more appropriate for understanding patterns within the available registration population.

Electric range analysis also revealed an important data-quality consideration: a large number of vehicles have a recorded electric range of zero. Consequently, positive-range observations should be analyzed separately when studying actual driving-range characteristics.

Overall, the project demonstrates that EV registrations vary considerably by manufacturer, vehicle type, model year and geographic location. The findings can help identify dominant manufacturers, popular vehicle categories and changing characteristics of the registered EV population. Future analysis could incorporate additional states, charging-station availability, population size, vehicle prices, incentives and time-series registration data to produce a more comprehensive measure of EV adoption.

## Recommendations

1. Expand geographic coverage-A nationwide dataset should be used if the goal is to compare EV adoption between states.
2. Normalize registration counts-Instead of comparing raw registration counts, calculate EV registrations per 1,000        residents.This would produce a fairer geographic comparison.
3. Separate zero-range observations-Zero electric range should be investigated before calculating range statistics.
4. Analyze BEV and PHEV separately-Their technology and range characteristics are different, so combining them can hide    important patterns.
5. Add charging infrastructure-Combining EV registration data with charging-station locations could reveal whether         charging infrastructure is associated with EV concentration.
6. Perform time-series analysis-If registration dates are available, analyze monthly or yearly growth instead of           relying only on model year.
7. Add economic variables-Vehicle price, income and incentives could provide a better explanation of EV adoption           patterns.
